Random sampling with numpy

If we need random data for ML/DL projects like otp generation we can use random module

```python
import numpy as np
import random 
```

To create random values we need to use .random and it will only give values between 0 to 1

```python
np.random.random(1)
```

```python
np.random.random((3,3))
```

To get a value in single integer type we can use randint where 1 is starting point and 4 is ending point

```python
np.random.randint(1,4)
```

We can create 2-D array using randint by providing range and shape

```python
np.random.randint(1,4,(4,4))
```

Here for creating 3-D array we have given value as 2 for creating two arrays

```python
np.random.randint(1,4,(2,4,4))
```

when we use randint it gives us diff values all the time but if we want to use same values then we can use seed function.

```python
np.random.seed(10)
np.random.randint(1,4,(2,4,4))
```

```python
np.random.seed(10)
np.random.randint(1,4,(2,4,4))
```

We can create N-D array using rand function

```python
np.random.rand(3)
```

```python
np.random.rand(3,3)
```

randn gives us negative values

```python
np.random.randn(3,3)
```

```python
np.random.choice(x)
```
