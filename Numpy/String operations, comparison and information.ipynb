String operations, comparison and information

If we have data in string type and we want to perform numpy operations on it then we can use string operations

```python
import numpy as np
```

```python
s1="Indian AI producion"
s2="Learning python numpy"
```

Numpy has 'char' module and char has 'add' method

```python
np.char.add(s1,s2)
```

```python
np.char.lower(s1)
```

```python
np.char.upper(s2)
```

```python
np.char.center(s2,60)
```

```python
np.char.center(s2,60,fillchar="*")
```

```python
np.char.split(s1)
```

When we have "\n" in our sentence and if we want to split it then we can use "splitlines"

```python
np.char.splitlines("hello\nIndian")
```

If we want to join a particular character with our string then we can use join method

```python
str3="dmy"
str4="dmy"
```

```python
np.char.join((":","/"),[str3,str4])
```

```python
np.char.join((":","/"),[s1,s2])
```

If we want to replace a word we can use replace function

```python
np.char.replace(s1,"AI","Artificial Intelligence")
```

If we use wrong string then it wont work

```python
np.char.replace(s1,"AIIIII","Artificial Intelligence")
```

To compare two string we can use equal method

```python
np.char.equal(str3,str4)
```

If we want to count the character present in the string we can use count method

```python
np.char.count(s1,"a")
```

Capital "A" and small "a" are counted differently here

```python
s1
```

if we use find method then We will get the index value of our word

```python
np.char.find(s1,"AI")
```
