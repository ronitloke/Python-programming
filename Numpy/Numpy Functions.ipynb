Numpy Functions

```python
import numpy as np
```

arange() function provides a range syntax arange(start, end, step)

```python
arr_1d=np.arange(1,13)
print(arr_1d)
```

To create even array provide step value as 1

```python
even_ar=np.arange(0,13,2)
print(even_ar)
```

using linspace() we can get difference between values

```python
np.linspace(1,20,3)
```

We can convert arrays into another dimension arrays using reshape(). Here above array was in one dimension and now it is converted into two dimension array

```python
arr_2d=arr_1d.reshape(2,6)
print(arr_2d)
```

```python
arr_2d=arr_1d.reshape(3,4)
print(arr_2d)
```

To convert 1d array into 3d array we need to provide values in sequential manner eg obj.reshape(3d value, 2d value, 1d value), also need to make sure the dot product of the array should match with the number of values present in 1d array.

```python
arr_3d=arr_1d.reshape(2,3,2)
print(arr_3d)
```

```python
arr_11d=np.arange(1,19)
print(arr_1d)
```

```python
arr_3d=arr_11d.reshape(2,3,3)
print(arr_3d)
```

```python
ar=np.arange(1,13).reshape(2,6)
print(ar)
```

ravel() converts multidimension array into one dimension array

```python
ar.ravel()
```

flatten() converts multidimension array into one dimension array but we can get output in certain orders by using order="C","F","A","K"

```python
ar.flatten()
```

```python
ar.flatten(order="C")
```

```python
ar.flatten(order="F")
```

```python
ar.flatten(order="A")
```

```python
ar.flatten(order="K")
```

transpose() converts rows into columns here rows are 6 and columns are 2 became opposite of wht we had up

```python
ar.transpose()
```

WE can even use obj.T for making use of transpose function
