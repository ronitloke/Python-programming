Mathematic operation using numpy function

```python
import numpy as np
```

```python
arr1=np.arange(1,10).reshape(3,3)
arr2=np.arange(1,10).reshape(3,3)
print(arr1)
print(arr2)
```

for adding two arrays use plus operator

```python
arr1 + arr2
```

Another way

```python
np.add(arr1,arr2)
```

For subtracting

```python
arr1-arr2
```

```python
np.subtract(arr1,arr2)
```

For dividing

```python
arr1/arr2
```

```python
np.divide(arr1,arr2)
```

For multiplication

```python
arr1*arr2
```

```python
np.multiply(arr1,arr2)
```

To find product of matrix i.e first rwo first column use @

```python
arr1@arr2
```

```python
arr1.dot(arr2)
```

To find maximum value

```python
arr1.max()
```

To find index position of maximum value

```python
arr1.argmax()
```

To find maximum values in row or column we can use axis value where 0 is column and 1 is row

```python
arr1.max(axis=0)
```

```python
arr1.max(axis=1)
```

```python
To find minimum value
```

```python
arr1.min()
```

```python
arr1.argmin()
```

```python
arr1.min(axis=0)
```

```python
arr1.min(axis=1)
```

To add all values from the matrix we can use sum()

```python
arr1.sum()
```

```python
np.sum(arr1)
```

To get sum of column use axis

```python
np.sum(arr1, axis=0)
```

To get sum of row use axis

```python
np.sum(arr1,axis=1)
```

To get average of matrix use min()

```python
np.mean(arr1)
```

```python
np.sqrt(arr1)
```

```python
To fine standard deviation use std()
```

```python
np.std(arr1)
```

To find exponent

```python
np.exp(arr1)
```

To find log

```python
np.log(arr1)
```

To find log with base 10

```python
np.log10(arr1)
```
