Array concatenation & split

```python
import numpy as np
```

```python
arr1=np.arange(1,17).reshape(4,4)
print(arr1)
```

```python
arr2=np.arange(17,33).reshape(4,4)
print(arr2)
```

```python
l1=[2,4,6,7]
l2=[1,8,3,5]
```

```python
l1+l2
```

we cannot use plus symbol as it add two arrays

```python
arr1+arr2
```

For concatenating column wise we can use this

```python
np.concatenate((arr1,arr2))
```

For concatenating row wise we can use axis=1

```python
np.concatenate((arr1,arr2),axis=1)
```

for concatenating column wise we can also use vstack which is vertical stack

```python
np.vstack((arr1,arr2))
```

for concatenating row wise we can also use hstack which is horizontal stack

```python
np.hstack((arr1,arr2))
```

```python
np.hstack((arr1,arr2,arr1))
```

for spliting array in parts we need to provide that value and we get value in list format

```python
np.split(arr1,2)
```

To split row wise we can provide axis value

```python
np.split(arr1,2,axis=1)
```

```python
np.split(arr1,4,axis=1)
```
