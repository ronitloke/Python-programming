Numpy Matrix

```python
import numpy as np
```

Creating a matrix

```python
mx_1s=np.array([[1,1,1],[1,1,1],[1,1,1]])
print(mx_1s)
```

If we want to create a matrix of 1000+ rows and columns then we can use obj.ones function and can provide columns value

```python
mx_1s=np.ones(5)
print(mx_1s)
```

```python
mx_1s.dtype
```

To create a matrix of rows and columns use obj.ones and give values in tuple (row,column)

```python
mx_1s=np.ones((3,4))
print(mx_1s)
```

till now we are getting values in float and if we want to get value in int we need to provide dtype=int

```python
mx_1s=np.ones((3,4),dtype=int)
print(mx_1s)
```

if we want all values as zeros then we need to use obj.zeros

```python
mx_0s=np.zeros((4,6))
print(mx_0s)
```

if we need boolean values we can provide dtype as bool and 0s represent false and 1s represent true

```python
mx_0s=np.zeros((4,6),dtype=bool)
print(mx_0s)
```

```python
mx_1s=np.ones((3,4),dtype=bool)
print(mx_1s)
```

```python
mx_1s=np.ones((3,4),dtype=str)
print(mx_1s)
```

If we try to convert zero values into str then we get empty string

```python
mx_0s=np.zeros((4,6),dtype=str)
print(mx_0s)
```

We can use bool as it will convert empty string to boolean value

```python
empty_str=""
print(bool(empty_str))
```

to create an empty matrix we can use obj.empty but output will not necessarily contain zero values

```python
em_mx=np.empty((3,3))
print(em_mx)
```
