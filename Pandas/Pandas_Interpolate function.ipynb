It will fill column according to the limit values provided, HEre 1 i provided so will fill 1 data in column

```python
df.interpolate(limit=1)
```

```python
df.interpolate(limit=2)
```

Here backward values are filled with ref to data present below nd it is filled accoring to limits provided

```python
df.interpolate(limit=1, limit_direction="backward")
```

```python
df.interpolate(limit=2, limit_direction="backward")
```

If we use both it will fill both values 1 of forward and other of backward

```python
df.interpolate(limit=1, limit_direction="both")
```

Only fill NaNs surrounded by valid values

```python
df.interpolate(limit_area="inside")
```

Only fill NaNs outside valid values (extrapolate)

```python
df.interpolate(limit_area="outside")
```

It will update the original dataframe but we need to call seperately

```python
df.interpolate(inplace=True)
df
```
