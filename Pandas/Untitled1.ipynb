How to create python pandas DataFrame

```python
import pandas as pd
```

```python
df=pd.DataFrame()
print(df)
```

```python
li=['a','b','c','d']
print(li)
```

```python
df=pd.DataFrame(li)
print(df)
```

```python
ml=[[1,2,3],[4,5,6],[7,8,9]]
print(ml)
```

```python
df1=pd.DataFrame(ml)
```

```python
dict_1={"ID":[1,2,3,4]}
print(dict_1)
```

```python
df2=pd.DataFrame(dict_1)
```
