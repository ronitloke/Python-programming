Pandas Merge Function

Pandas merge function connects columns or indexes in dataframe based on one or more keys.

```python
import pandas as pd
```

```python
df1=pd.DataFrame({"ID":[1,2,3,4],
                "Class":[9,10,11,12]})
df1
```

```python
df2=pd.DataFrame({"ID":[1,2,3,5],
                "Name":["A","B","C","D"]})
df2
```

For merging two columns we need to use pd.merge() in which "on" is a parameter where we can use label for single key and list for multiple keys

```python
pd.merge(df1,df2,on="ID")
```

```python
pd.merge(df2,df1,on="ID")
```

Here only 3 rows are printed bcaz only 3 rows are similar

```python
pd.merge(df1,df2,on="ID")
```

```python
pd.merge(df1,df2,on="ID", how="inner")
```

It will take according to first dataframe since ID=5 is not there in df1 so NaN is printed is name

```python
pd.merge(df1,df2,on="ID", how="left")
```

```python
pd.merge(df1,df2,on="ID", how="right")
```

If we want to print all values we need to take how=outer

```python
pd.merge(df1,df2,on="ID", how="outer")
```

If we want to know which values belongs to which class then we need to use indicator=True

```python
pd.merge(df1,df2,on="ID", how="outer", indicator=True)
```

```python
df2=pd.DataFrame({"ID":[5,6,7,8],
                "Name":["A","B","C","D"]})
df2
```

If we want to show both index then use this. Usually used when we have multiple index to show

```python
pd.merge(df1,df2,left_index=True,right_index=True)
```

```python
pd.merge(df1,df2)
```

Now we will check what if we have both dataframe with same class data

```python
df1=pd.DataFrame({"ID":[1,2,3,4],
                "Class":[9,10,11,12]})
df1
```

```python
df2=pd.DataFrame({"ID":[1,2,3,4],
                "Class":[9,10,11,12]})
df2
```

```python
pd.merge(df1,df2,on="ID")
```

We can use suffixes for overlapping columns and can provide our names but need to provide values in tuple form

```python
pd.merge(df1,df2,on="ID", suffixes=("_Higher","_Middle"))
```
