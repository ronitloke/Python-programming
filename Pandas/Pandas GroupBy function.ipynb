Pandas GroupBy function

Pandas groupby function is used to split into groups based on some criteria

```python
import pandas as pd
```

```python
df=pd.read_csv("D:\Programming\Python programming\Pandas\Student_results2.csv")
df
```

We can use this to split a column and store it in another variable

```python
gr1=df.groupby(by="Section")
gr1
```

Multiple values are taken of column (section) and merged into single group and index values are displayed

```python
gr1.groups
```

WE can group multiple columns as well

```python
df.groupby(["Section","Class"]).groups
```

It shows the keys and values present in dataframe where section is the key and df_1 is value

```python
for Section, df_1 in gr1:
    print(Section)
    print(df_1)
```

```python
list(gr1)
```

```python
dict(list(gr1))
```

```python
gr3=df.groupby("Class").get_group(10)
gr3
```

```python
gr3=df.groupby("Section").get_group("A")
gr3
```
