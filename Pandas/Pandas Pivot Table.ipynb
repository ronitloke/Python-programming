Pandas Pivot Table

The levels in the pivot table will be stored in MultiIndex objects(Hierarchical indexes) on the index and column of result dataframe.

```python
import pandas as pd
```

```python
df=pd.read_csv("D:\Programming\Python programming\Pandas\Movie_ratings.csv")
df
```

The output shows the aggregrate value i.e the mean value 

```python
df.pivot_table(index="Genre")
```

We can get information according to the columns mentioned

```python
df.pivot_table(index="Genre", columns="Year")
```

To get the count of values we can use aggfunc="count"

```python
df.pivot_table(index="Genre", columns="Year", aggfunc="count")
```

```python
df.pivot_table(index="Genre", columns="Year", aggfunc="max")
```

```python
df.pivot_table(index="Genre", columns="Year", aggfunc="sum")
```

If we want to fill the missing values we can use fill_value

```python
df.pivot_table(index="Genre", columns="Year", fill_value=0)
```

Margins will give grand total i.e it will add all rows and columns

```python
df.pivot_table(index="Genre", columns="Year", fill_value=0, margins=True)
```

```python
df.pivot_table(index="Genre", columns="Year", aggfunc="sum", fill_value=0, margins=True)
```
