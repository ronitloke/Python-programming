Pandas loc 

```python
import pandas as pd
```

```python
df=pd.read_csv("D:\Programming\Python programming\Pandas\Student_results1.csv")
df
```

It will return the values present in the rows

```python
df.loc[0]
```

```python
df.loc[4]
```

For multiple rows

```python
df.loc[[0,4]]
```

We can access specific value present in a column by using df.loc[index,column]

```python
df.loc[4,"Class"]
```

by using slicing method

```python
df.loc[0:3,"Class"]
```

```python
df.loc[0:4,"percentage"]
```

It is used to display which rows we want to print and which we dont. Need to provide exact no of booleean values for the rows present in dataset eg: df.loc[[boolean values]]

```python
df.loc[[True, False, True, False,True]]
```

It will print values less than 11 present in class column eg:df.loc[df[column] < value]

```python
df.loc[df["Class"]<11]
```

It will only print percentage values less than 11 as per class column

```python
df.loc[df["Class"]<11,"percentage"]
```
