Pandas FillNA

```python
import pandas as pd
```

```python
df=pd.read_csv("D:\Programming\Python programming\Pandas\Fortune_10.csv")
df
```

We can fill missing values with help of fill na by providing values in its parameters

```python
df.fillna(1)
```

```python
df.fillna({"Name":"none","Industry":"nothing"})
```

Method is use when we need to fill the values which are present in forward or backward index

```python
df.fillna(method="ffill")
```

```python
df.ffill()
```

```python
df.bfill()
```

It fills the values in next rows from values present in previous rows 

```python
df.ffill(axis=0)
```

It fills the values in previous columns from values present on next column

```python
df.bfill(axis=1)
```
