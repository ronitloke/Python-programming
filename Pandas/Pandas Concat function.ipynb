Pandas Concat function

It is used to combine series, dataframe and panel objects

```python
import pandas as pd
```

```python
sr1=pd.Series([0,1,2])
sr1
```

```python
sr2=pd.Series([3,4,5,6,7])
sr2
```

 series, dataframe or panel objects whatever we use we need to give inputs in list in concat function. We can even concat two uneven series

```python
pd.concat([sr1,sr2])
```

```python
df1=pd.DataFrame({"ID":[1,2,3,4],
                 "Name":["A","B","C","D"],
                 "Class":[5,6,7,8]})
df1
```

```python
df2=pd.DataFrame({"ID":[5,6,7,8],
                 "Name":["E","F","G","H"],
                 "Class":[9,10,11,12]})
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1])
```

If we need to see data column wise then we can provide 1 as axis value

```python
pd.concat([df1,df2], axis=1)
```

```python
pd.concat([df1,df2], axis=0)
```

To get the index number systematically we can use ignore_index

```python
pd.concat([df1,df2], axis=0, ignore_index=True)
```

```python
df1=pd.DataFrame({"ID":[1,2,3,4],
                 "Name":["A","B","C","D"],
                 "Class":[5,6,7,8]})
df1
```

```python
df2=pd.DataFrame({"ID":[7,8],
                 "Name":["G","H"],
                 "Class":[11,12]})
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

If we use join ="inner" it will return only intersecting value rows

```python
pd.concat([df1,df2],axis=1, join="inner")
```

join="outer" will give output of all elements. It is union of both dataframe

```python
pd.concat([df1,df2],axis=1, join="outer")
```

When we use keys it provides labels to the dataframes which are merged

```python
df1=pd.DataFrame({"ID":[1,2,3,4],
                 "Name":["A","B","C","D"],
                 "Class":[5,6,7,8]})
df1
```

```python
df2=pd.DataFrame({"ID":[5,6,7,8],
                 "Name":["E","F","G","H"],
                 "Class":[9,10,11,12]})
df2
```

```python
pd.concat([df1,df2],keys=["df1","df2"])
```

```python
pd.concat([df1,df2],keys=["First df","Second df"])
```

```python
pd.concat([df1,df2],axis=1,keys=["First df","Second df"])
```
