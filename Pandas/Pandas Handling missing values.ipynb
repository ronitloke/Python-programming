```python
df=pd.read_csv("D:\Programming\Python programming\Pandas\Fortune_10.csv")
df
```

Drops the rows which has missing values

```python
df.dropna()
```

Drops the column which has missing values....defauld axis value is zero so it drops row

```python
df.dropna(axis=1)
```

drops rows and columns if they have NaN values

```python
df.dropna(how = "any")
```

only considers axis 

```python
df.dropna(axis=1, how="any")
```

Drops only when there are NaN values in all rows and columns

```python
df.dropna(how = "all")
```

Thresh is used when we need atleast one not null value in a row 

```python
df.dropna(thresh=1)
```

if we set thresh=2 then it will select rows with atleast two not null values

```python
df.dropna(thresh=2)
```
