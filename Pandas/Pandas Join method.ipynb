Pandas Join method

DataFrame join is a convenient method for combining the columns of two potentially differently-indexed.

```python
import pandas as pd
```

```python
df1=pd.DataFrame({"A":[1,2,3],
                "B":[10,20,30]}
                )
df2=pd.DataFrame({"C":[4,5,6],
                "D":[40,50,60]})

display(df1,df2)
```

```python
df1.join(df2)
```

```python
df1=pd.DataFrame({"A":[1,2,3],
                "B":[10,20,30]},
                 index=["a","b","c"]
                )
df2=pd.DataFrame({"C":[4,5,6],
                "D":[40,50,60]}
                )

display(df1,df2)
```

It is not considering df2 values as index is not similar/


```python
df1.join(df2)
```

```python
df1=pd.DataFrame({"A":[1,2,3],
                "B":[10,20,30]},
                 index=["a","b","c"]
                )
df2=pd.DataFrame({"C":[4,5,6],
                "D":[40,50,60]}
                )

display(df1,df2)
```

```python
df1=pd.DataFrame({"A":[1,2,3],
                "B":[10,20,30]},
                 index=["a","b","c"]
                )
df2=pd.DataFrame({"C":[4,5],
                "D":[40,50]},
                index=["a","b"]
                )

display(df1,df2)
```

```python
df1.join(df2)
```

If there is missing data then we can use these parameters to get output accordingly

```python
df1.join(df2,how="right")
```

```python
df1.join(df2,how="inner")
```

```python
df1.join(df2,how="outer")
```

```python
df1=pd.DataFrame({"A":[1,2,3],
                "B":[10,20,30]},
                 index=["a","b","c"]
                )
df2=pd.DataFrame({"A":[4,5],
                "D":[40,50]},
                index=["a","b"]
                )

display(df1,df2)
```

If we have two columns having same names and they are overlapping then we can use lsuffix and rsuffix to provide names

```python
df1.join(df2, lsuffix="_1")
```

```python
df1.join(df2, rsuffix="_1")
```
