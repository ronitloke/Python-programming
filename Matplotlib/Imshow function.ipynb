How to show image using matplotlib imshow function?

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

To read a image we can use imread() function

```python
img=mpimg.imread("Pie_chart.png")
```

It stores data in numeric format

```python
img
```

```python
type(img)
```

```python
img.shape
```

```python
img.ndim
```

To display the image we can use imshow() function

```python
plt.imshow(img)
plt.show()
```

```python
plt.figure(figsize=(16,9))
plt.axis("off")
plt.imshow(img)
plt.colorbar()
plt.show()
```

cmap is color map

```python
plt.figure(figsize=(16,9))
plt.axis("off")
plt.imshow(img, cmap="hot")
plt.colorbar()
plt.show()
```

We can get image data according to channel wise

```python
single_channel=img[:,:,1]
plt.figure(figsize=(16,9))
plt.axis("off")
plt.imshow(img, cmap="hot")
plt.colorbar()
plt.show()
```

```python
img2=mpimg.imread("Nature_celebrating_India.png")
plt.imshow(img2)
plt.figure(figsize=(16,9))
plt.axis("off")
plt.show()
```

```python
single_channel2=img2[:,:,1]

img2=mpimg.imread("Nature_celebrating_India.png")
plt.imshow(single_channel2)
plt.figure(figsize=(16,9))
plt.axis("off")
plt.savefig("Nature.png")
plt.show()
```
