How to save figures (Plots and Charts)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import random
from matplotlib import style
```

If we want to save a figure we can use savefig() function it will save in the folder where we are saving code in PNG format by default

```python
plt.pie([40,30,20])
plt.savefig("Pie_chart")
plt.show()
```

We can set dpi which is dot per inch for resolution

```python
plt.pie([40,30,20])
plt.savefig("Pie_chart", dpi=100, quality=99, facecolor="g")
plt.show()
```
