Matplotlib Barchart

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
# Dataset of 'Indian Artificial Intelligence Production (IAIP) Class"
 
#classes = ["Python", "R", "Artificial Intelligence", "Machine Learning", "Data Science"]
 
classes = ["Python", "R", "AI", "ML", "DS"]
class1_students = [30, 10, 20, 25, 10] # out of 100 student in each class
class2_students = [40, 5, 20, 20, 10]
class3_students = [35, 5, 30, 15, 15]
```

For plotting a bar chart we need to use plt.bar(x-axis,y-axis) function. Here we want to show classes on x-axis and class1 on y-axis

```python
plt.bar(classes,class1_students)
```

for plotting bar horizontally we can use barh function

```python
plt.barh(classes,class1_students)
```

```python
plt.bar(classes,class1_students, width =0.2)
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge")
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y")
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m")
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m", linewidth=5)
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m", linewidth=5, alpha=0.5)
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m", linewidth=5, alpha=0.5, linestyle="--")
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m", linewidth=5, alpha=0.5,
        linestyle="--",label="class1_students")
plt.legend()
```

```python
plt.bar(classes,class1_students, width =0.2,align="edge", color="y", edgecolor="m", linewidth=5, alpha=0.5,
        linestyle="--",label="class1_students",visible=False)
```

```python
style.use("ggplot")
plt.bar(classes,class1_students, width =0.6,align="edge", color="k", edgecolor="m", linewidth=5, alpha=0.9,
        linestyle="--",label="class1_students")
```

We need to provide figsize value in tuple

```python
style.use("ggplot")
plt.figure(figsize=(16,9))
plt.bar(classes,class1_students, width =0.6,align="edge", color="k", edgecolor="m", linewidth=5, alpha=0.9,
        linestyle="--",label="class1_students")
```

Once we use style.use we dont need to use it again and again

```python
plt.figure(figsize=(16,9))
plt.barh(classes,class1_students,align="edge", color="k", edgecolor="m", linewidth=5, alpha=0.9,
        linestyle="--",label="class1_students")
```

```python
plt.figure(figsize=(16,9))
plt.barh(classes,class1_students,align="edge", color="k", edgecolor="m", linewidth=5, alpha=0.9,
        linestyle="--",label="class1_students")
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

plt.bar(classes,class1_students,width=0.2,color="b",label="class1_students")
plt.bar(classes,class2_students,width=0.2,color="g",label="class2_students")

plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

Due to overlapping of index we are using seperate class index

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2
plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index,class2_students,width,color="g",label="class2_students")

plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

still it is overlapping bcaz two classes are having same spots so we need to create a difference by adding width in class2 students

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")

plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

But still it was not showing label for classes it was showing index no spo for tht we need to use xticks

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")

plt.xticks(classes_index + width, classes)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")

plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")
plt.bar(classes_index + width,class3_students,width,color="y",label="class3_students")


plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

Here again class 2 and 3 are overlapping so solution for this would be to do width + width in class 3. 

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")
plt.bar(classes_index + width + width,class3_students,width,color="y",label="class3_students")


plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

But we can use width only till 100% value we we try to go beyond that graphs will over lap like we r changing width value to 0.6 to see the difference

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.6

plt.bar(classes_index,class1_students,width,color="b",label="class1_students")
plt.bar(classes_index + width,class2_students,width,color="g",label="class2_students")
plt.bar(classes_index + width + width,class3_students,width,color="y",label="class3_students")


plt.xticks(classes_index + width, classes, rotation =20)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.xlabel("Classes",fontsize=15)
plt.ylabel("No of students",fontsize=15)
plt.show()
```

For plotting horizontal bar chart we need to change few things like barh, yticks,ylabel,xlabel.

```python
plt.figure(figsize=(16,9))

classes_index=np.arange(len(classes))
width=0.2

plt.barh(classes_index,class1_students,width,color="b",label="class1_students")
plt.barh(classes_index + width,class2_students,width,color="g",label="class2_students")
plt.barh(classes_index + width + width,class3_students,width,color="y",label="class3_students")


plt.yticks(classes_index + width, classes, rotation =20)
plt.title("Bar chart of IAIP class", fontsize=18)
plt.ylabel("Classes",fontsize=15)
plt.xlabel("No of students",fontsize=15)
plt.legend()
plt.show()
```
