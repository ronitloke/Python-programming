Matplotlib Histogram

```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
ml_students_age = np.random.randint(18,45, (100))
py_students_age = np.random.randint(15,40, (100))
```

```python
print(ml_students_age)
print(py_students_age)
```

We can plot histogram using with the help of hist function

```python
plt.hist(ml_students_age)
plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.show()
```

```python
If we want to define the category for Bar then we can use Bins and provide our desired values 
```

```python
bins=[15,20,25,30,35,40,45]

plt.hist(ml_students_age, bins)
plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.show()
```

If we need space btwn bars we can use rwidth

```python
bins=[15,20,25,30,35,40,45]

plt.hist(ml_students_age, bins, rwidth=0.8)
plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.show()
```

```python
bins=[15,20,25,30,35,40,45]

plt.hist(ml_students_age, bins, rwidth=0.8, histtype="step")
plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.show()
```

```python
bins=[15,20,25,30,35,40,45]

plt.hist(ml_students_age, bins, rwidth=0.8, orientation="vertical", color="m", label="ML Student")
plt.hist(py_students_age, bins, rwidth=0.8, orientation="horizontal", color="y", label="PY Student")

plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.legend()
plt.show()
```

If we want to show histogram properly then we need to display names in the list in same hist function along with color and label

```python
bins=[15,20,25,30,35,40,45]

plt.hist([ml_students_age,py_students_age], bins, rwidth=0.8, orientation="vertical", color=["m","y"], label=["ML Student","PY Student"])
#plt.hist(py_students_age, bins, rwidth=0.8, orientation="vertical", color="y", label="PY Student")

plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.legend()
plt.show()
```

For changing size of histogram we can use figure function

```python
bins=[15,20,25,30,35,40,45]
plt.figure(figsize=(16,9))

plt.hist([ml_students_age,py_students_age], bins, rwidth=0.8, orientation="vertical", color=["m","y"], label=["ML Student","PY Student"])

plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.legend()
plt.show()
```

```python
from matplotlib import style
style.use("ggplot")
plt.figure(figsize=(16,9))

plt.hist([ml_students_age,py_students_age], bins, rwidth=0.8, orientation="vertical", color=["m","y"], label=["ML Student","PY Student"])

plt.title("ML students age histogram")
plt.xlabel("Students age category")
plt.ylabel("No of students age")
plt.legend()
plt.show()
```
