```python
df=pd.read_csv("D:\Programming\Python programming\Matplotlib\googleplaystore.csv", nrows=1000)
df.shape
```

```python
df
```

```python
x=df["Rating"]
y=df["Reviews"]
```

For plotting a scatter plot graph we need to use plt.scatter() function

```python
plt.scatter(x,y)
```

```python
plt.scatter(x,y)
plt.title("Google play store scatter plot")
plt.xlabel("Ratings")
plt.ylabel("Reviews")
```

We can apply various parameters in scatter function to change marker size use s=100 and for transperancy use alpha and c is for color, verts work as marker.

```python
plt.figure(figsize=(16,9))
plt.scatter(x,y,c="r", marker="*", s=100, alpha=0.5, linewidth=10, edgecolor="g")
plt.title("Google play store scatter plot")
plt.xlabel("Ratings")
plt.ylabel("Reviews")
```

For getting two scatter plots in one graph

```python
plt.figure(figsize=(16,9))

plt.scatter(x,y,c="r", marker="*", s=100, alpha=0.5, linewidth=10, edgecolor="g")
plt.scatter(x,df["Installs"],c="y", marker="o", s=100, alpha=0.5, linewidth=10, edgecolor="c")

plt.title("Google play store scatter plot")
plt.xlabel("Ratings")
plt.ylabel("Reviews & Installs")
```
